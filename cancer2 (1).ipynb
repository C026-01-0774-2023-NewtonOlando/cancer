```python
import hashlib
from collections import defaultdict
from PIL import Image # Needed for image processing
from io import BytesIO # Needed by decode_image function

# The `decode_image` function and `df` DataFrame are available from previous cells.

# ============================================================
# 1. PREPARE IMAGE DATA FROM DATAFRAME
# ============================================================

# Create a list of dictionaries, each representing an image with its original index,
# decoded PIL image, and label. This replaces the file-based 'images' list.
df_images_info = []
for i in range(len(df)):
    image_data_dict = df.iloc[i]["image"]
    # Ensure consistent format for hashing by converting to RGB
    pil_image = decode_image(image_data_dict).convert("RGB")
    label = df.iloc[i]["label"]
    df_images_info.append({
        "original_index": i,
        "pil_image": pil_image,
        "label": label
    })

print("Total images found in DataFrame:", len(df_images_info))


# ============================================================
# 2. CALCULATE HASH FOR EVERY IMAGE
# ============================================================

hash_groups = defaultdict(list)

for image_info in df_images_info:
    try:
        # Get bytes from the PIL image and hash them
        image_hash = hashlib.md5(image_info["pil_image"].tobytes()).hexdigest()
        hash_groups[image_hash].append(image_info)
    except Exception as e:
        print(f"Error processing image at original index {image_info['original_index']}: {e}")

print("Unique image contents:", len(hash_groups))


# ============================================================
# 3. CLASSIFY DUPLICATES
# ============================================================

same_label_duplicates = []
different_label_duplicates = []

for image_hash, group in hash_groups.items():

    # Only interested in actual duplicates (groups with more than one image)
    if len(group) <= 1:
        continue

    # Check if all images in the group have the same label
    labels = set(image_info["label"] for image_info in group)

    if len(labels) > 1:
        # SAME IMAGE EXISTS IN DIFFERENT LABELS
        # All copies should typically be removed in this case
        different_label_duplicates.append(group)
    else:
        # SAME IMAGE EXISTS ONLY WITHIN ONE LABEL
        # Keep one, remove the rest
        same_label_duplicates.append(group)


# ============================================================
# 4. DISPLAY DUPLICATE SUMMARY
# ============================================================

print("\n================ DUPLICATE SUMMARY ================")

print(
    "Duplicate groups with DIFFERENT labels:",
    len(different_label_duplicates)
)

print(
    "Duplicate groups within SAME label:",
    len(same_label_duplicates)
)

print(
    "Total duplicate groups:",
    len(different_label_duplicates) + len(same_label_duplicates)
)


# ============================================================
# 5. COUNT IMAGES TO BE REMOVED
# ============================================================

different_label_images_to_remove = sum(
    len(group)
    for group in different_label_duplicates
)

same_label_images_to_remove = sum(
    len(group) - 1  # Keep one, remove the rest
    for group in same_label_duplicates
)

total_to_remove = (
    different_label_images_to_remove +
    same_label_images_to_remove
)

print("\nImages that will be removed (based on content hash):")
print("Different-label duplicates (all copies):", different_label_images_to_remove)
print("Same-label extra copies (all but one):", same_label_images_to_remove)
print("TOTAL TO BE REMOVED:", total_to_remove)


# ============================================================
# 6. SHOW DIFFERENT-LABEL DUPLICATES (for review)
# ============================================================

print("\n================ DIFFERENT-LABEL DUPLICATES (Content identical, labels vary) ================")

if not different_label_duplicates:
    print("No different-label duplicate groups found.")
else:
    for number, group in enumerate(different_label_duplicates, start=1):
        # Get the hash from the first image in the group for reference
        group_hash = hashlib.md5(group[0]["pil_image"].tobytes()).hexdigest()
        print(f"\nGroup {number} (Hash: {group_hash})")
        for image_info in group:
            print(
                f"  Original Index: {image_info['original_index']} | "
                f"Label: {image_info['label']}"
            )


# ============================================================
# 7. SHOW SAME-LABEL DUPLICATES (for review)
# ============================================================

print("\n================ SAME-LABEL DUPLICATES (Content identical, labels same) ================")

if not same_label_duplicates:
    print("No same-label duplicate groups found.")
else:
    for number, group in enumerate(same_label_duplicates, start=1):
        # Get the hash from the first image in the group for reference
        group_hash = hashlib.md5(group[0]["pil_image"].tobytes()).hexdigest()
        print(f"\nGroup {number} (Hash: {group_hash})")
        # The first image in the group can be considered the 'keeper' for display purposes
        for i, image_info in enumerate(group):
            status = "(keeper)" if i == 0 else "(remove)"
            print(
                f"  Original Index: {image_info['original_index']} | "
                f"Label: {image_info['label']} {status}"
            )
```

```python
import pandas as pd
import os

# ============================================================
# BUILD THE CLEANED DATAFRAME AND REMOVAL REPORT
# ============================================================

kept_original_indices = []
removed_records = [] # To store dictionaries for the removed_df

# hash_groups is available from the previous cell execution
for image_hash, group in hash_groups.items():
    # 'group' is a list of dictionaries, each containing
    # 'original_index', 'pil_image', and 'label'

    # Check labels within the group to determine if they conflict
    labels_in_group = set(item['label'] for item in group)

    if len(group) > 1: # This is a duplicate group (more than one image with this hash)
        if len(labels_in_group) > 1:
            # Case 1: Same image content appears with DIFFERENT LABELS
            # All images in this group are considered problematic and will be removed.
            for image_info in group:
                removed_records.append({
                    "original_index": image_info["original_index"],
                    "label": image_info["label"],
                    "reason": "Duplicate image content, conflicting labels"
                })
        else:
            # Case 2: Same image content appears multiple times but with the SAME LABEL
            # We keep the first instance and remove the rest as redundant copies.
            keeper_image_info = group[0]
            kept_original_indices.append(keeper_image_info["original_index"])

            for duplicate_image_info in group[1:]:
                removed_records.append({
                    "original_index": duplicate_image_info["original_index"],
                    "label": duplicate_image_info["label"],
                    "reason": "Duplicate image content, same label (extra copy)"
                })
    else:
        # This group contains only one image (unique content), so it is kept.
        kept_original_indices.append(group[0]["original_index"])

# Create the cleaned DataFrame by selecting rows from the original df
# using the indices of the images we decided to keep.
cleaned_df = df.iloc[kept_original_indices].reset_index(drop=True)

# Create a DataFrame for the removed items, which will serve as a report.
removed_df = pd.DataFrame(removed_records)

# ============================================================
# SAVE REMOVAL REPORT (as a CSV file)
# ============================================================

# Define a path for the removal report in the current content directory.
report_filename = "removed_duplicates_report.csv"
report_path = os.path.join("/content", report_filename)

if not removed_df.empty:
    removed_df.to_csv(report_path, index=False)
    print(f"\nRemoval report saved to: {report_path}")
else:
    print("\nNo duplicate images were removed, so no removal report was generated.")

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n===============================================")
print("CLEANING COMPLETE")
print("===============================================")

print("Original images (in DataFrame):", len(df))
print("Images kept (in cleaned_df):", len(cleaned_df))
print("Images removed (as per report):", len(removed_records))

# Perform a consistency check to ensure counts add up correctly.
expected_total = len(cleaned_df) + len(removed_records)
if len(df) == expected_total:
    print(f"Consistency check: Original count ({len(df)}) = Kept ({len(cleaned_df)}) + Removed ({len(removed_records)}). Consistent.")
else:
    print(f"Consistency check: WARNING - Original count ({len(df)}) != Kept ({len(cleaned_df)}) + Removed ({len(removed_records)}). Inconsistent.")

print("\nCleaned DataFrame (first 5 rows):")
display(cleaned_df.head())

# The original 'dataset_path', 'cleaned_path', and 'removed_path' are not needed
# for this in-memory DataFrame-based approach, so they are not used.
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        # Clear previous gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update model weights
        optimizer.step()

        # Track loss
        running_loss += loss.item() * images.size(0)

        # Track accuracy
        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_accuracy = correct / total

    return epoch_loss, epoch_accuracy
```

```python
def validate(model, loader, criterion, device):
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    # No gradients needed during validation
    with torch.no_grad():

        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            # Forward pass
            outputs = model(images)

            # Calculate loss
            loss = criterion(outputs, labels)

            # Track loss
            running_loss += loss.item() * images.size(0)

            # Track accuracy
            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_accuracy = correct / total

    return epoch_loss, epoch_accuracy
```

```python
import copy
import time
import torch

# Number of epochs
num_epochs = 15

# Store training history
history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": []
}

# Keep a copy of the best model
best_model_weights = copy.deepcopy(model.state_dict())
best_val_loss = float("inf")

# Early stopping settings
patience = 5
epochs_without_improvement = 0

print("Starting training...")
print(f"Device: {device}")
print(f"Epochs: {num_epochs}")
print("-" * 60)

for epoch in range(num_epochs):

    start_time = time.time()

    # -------------------------
    # TRAINING
    # -------------------------
    train_loss, train_accuracy = train_one_epoch(
        model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    # -------------------------
    # VALIDATION
    # -------------------------
    val_loss, val_accuracy = validate(
        model,
        val_loader,
        criterion,
        device
    )

    # Save history
    history["train_loss"].append(train_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_loss"].append(val_loss)
    history["val_accuracy"].append(val_accuracy)

    # Calculate time
    epoch_time = time.time() - start_time

    # Print results
    print(
        f"Epoch [{epoch + 1}/{num_epochs}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f} | "
        f"Time: {epoch_time:.1f}s"
    )

    # -------------------------
    # SAVE BEST MODEL
    # -------------------------
    if val_loss < best_val_loss:

        best_val_loss = val_loss
        best_model_weights = copy.deepcopy(model.state_dict())

        epochs_without_improvement = 0

        print("  ✓ New best model saved in memory.")

    else:
        epochs_without_improvement += 1

    # -------------------------
    # EARLY STOPPING
    # -------------------------
    if epochs_without_improvement >= patience:

        print(
            f"\nEarly stopping triggered after "
            f"{patience} epochs without validation improvement."
        )

        break

# Restore the best model
model.load_state_dict(best_model_weights)

print("\n" + "=" * 60)
print("TRAINING COMPLETE")
print("=" * 60)

print(f"Best validation loss: {best_val_loss:.4f}")
print(f"Best epoch: {history['val_loss'].index(best_val_loss) + 1}")
```

```python
import torch
import torch.nn as nn

class PatchEmbedding(nn.Module):
    def __init__(self, image_size=224, patch_size=16, in_channels=3, embed_dim=128):
        super().__init__()

        self.image_size = image_size
        self.patch_size = patch_size
        self.num_patches = (image_size // patch_size) ** 2

        self.projection = nn.Conv2d(
            in_channels=in_channels,
            out_channels=embed_dim,
            kernel_size=patch_size,
            stride=patch_size
        )

    def forward(self, x):
        # x shape:
        # [batch, 3, 224, 224]

        x = self.projection(x)

        # [batch, 128, 14, 14]
        x = x.flatten(2)

        # [batch, 128, 196]
        x = x.transpose(1, 2)

        # [batch, 196, 128]
        return x


# Test our patch embedding
patch_embedding = PatchEmbedding()

test_images, test_labels = next(iter(vit_train_loader))

patches = patch_embedding(test_images)

print("Input shape:       ", test_images.shape)
print("Patch output shape:", patches.shape)
print("Number of patches:", patch_embedding.num_patches)
```

```python
class AddCLSAndPosition(nn.Module):
    def __init__(self, num_patches=196, embed_dim=128):
        super().__init__()

        # Learnable CLS token
        self.cls_token = nn.Parameter(
            torch.zeros(1, 1, embed_dim)
        )

        # Learnable positional embeddings
        # +1 because of the CLS token
        self.position_embedding = nn.Parameter(
            torch.zeros(1, num_patches + 1, embed_dim)
        )

    def forward(self, x):
        # x shape:
        # [batch, 196, 128]

        batch_size = x.size(0)

        # Expand CLS token for the whole batch
        cls_token = self.cls_token.expand(
            batch_size, -1, -1
        )

        # Add CLS token at the beginning
        x = torch.cat(
            (cls_token, x),
            dim=1
        )

        # Add positional information
        x = x + self.position_embedding

        return x


# Create the component
cls_position = AddCLSAndPosition(
    num_patches=patch_embedding.num_patches,
    embed_dim=128
)

# Test it
tokens = cls_position(patches)

print("Before CLS token: ", patches.shape)
print("After CLS token:  ", tokens.shape)
print("CLS token shape:  ", cls_position.cls_token.shape)
print("Position shape:   ", cls_position.position_embedding.shape)
```

```python
class MultiHeadSelfAttention(nn.Module):
    def __init__(self, embed_dim=128, num_heads=4, dropout=0.1):
        super().__init__()

        # Make sure embedding dimension can be divided
        # equally among the attention heads
        assert embed_dim % num_heads == 0

        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads

        # Linear layers for Query, Key and Value
        self.query = nn.Linear(embed_dim, embed_dim)
        self.key = nn.Linear(embed_dim, embed_dim)
        self.value = nn.Linear(embed_dim, embed_dim)

        # Final projection after attention
        self.output_projection = nn.Linear(
            embed_dim,
            embed_dim
        )

        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x:
        # [batch_size, sequence_length, embed_dim]

        batch_size, sequence_length, _ = x.shape

        # Create Q, K and V
        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        # Split embedding dimension into multiple heads
        Q = Q.view(
            batch_size,
            sequence_length,
            self.num_heads,
            self.head_dim
        ).transpose(1, 2)

        K = K.view(
            batch_size,
            sequence_length,
            self.num_heads,
            self.head_dim
        ).transpose(1, 2)

        V = V.view(
            batch_size,
            sequence_length,
            self.num_heads,
            self.head_dim
        ).transpose(1, 2)

        # Attention scores
        attention_scores = torch.matmul(
            Q,
            K.transpose(-2, -1)
        )

        # Scale scores
        attention_scores = attention_scores / (
            self.head_dim ** 0.5
        )

        # Convert scores into probabilities
        attention_weights = torch.softmax(
            attention_scores,
            dim=-1
        )

        attention_weights = self.dropout(
            attention_weights
        )

        # Combine values according to attention weights
        attention_output = torch.matmul(
            attention_weights,
            V
        )

        # Put the heads back together
        attention_output = attention_output.transpose(
            1, 2
        ).contiguous()

        attention_output = attention_output.view(
            batch_size,
            sequence_length,
            self.embed_dim
        )

        # Final projection
        output = self.output_projection(
            attention_output
        )

        return output


# Create our attention module
attention = MultiHeadSelfAttention(
    embed_dim=128,
    num_heads=4
)

# Test it
attention_output = attention(tokens)

print("Input token shape:   ", tokens.shape)
print("Attention output:    ", attention_output.shape)
print("Number of heads:     ", attention.num_heads)
print("Dimensions per head: ", attention.head_dim)
```

```python
class TransformerEncoderBlock(nn.Module):
    def __init__(
        self,
        embed_dim=128,
        num_heads=4,
        mlp_dim=256,
        dropout=0.1
    ):
        super().__init__()

        # First LayerNorm
        self.norm1 = nn.LayerNorm(embed_dim)

        # Multi-Head Self-Attention
        self.attention = MultiHeadSelfAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout
        )

        # Second LayerNorm
        self.norm2 = nn.LayerNorm(embed_dim)

        # MLP / Feed-Forward Network
        self.mlp = nn.Sequential(
            nn.Linear(embed_dim, mlp_dim),
            nn.GELU(),
            nn.Dropout(dropout),

            nn.Linear(mlp_dim, embed_dim),
            nn.Dropout(dropout)
        )

    def forward(self, x):

        # ------------------------------------------------
        # 1. Self-Attention + Residual Connection
        # ------------------------------------------------

        attention_output = self.attention(
            self.norm1(x)
        )

        x = x + attention_output

        # ------------------------------------------------
        # 2. MLP + Residual Connection
        # ------------------------------------------------

        mlp_output = self.mlp(
            self.norm2(x)
        )

        x = x + mlp_output

        return x


# Create one Transformer Encoder Block
encoder_block = TransformerEncoderBlock(
    embed_dim=128,
    num_heads=4,
    mlp_dim=256,
    dropout=0.1
)

# Test it
encoder_output = encoder_block(tokens)

print("Input shape:  ", tokens.shape)
print("Output shape: ", encoder_output.shape)
```

```python
class TransformerEncoder(nn.Module):
    def __init__(
        self,
        depth=4,
        embed_dim=128,
        num_heads=4,
        mlp_dim=256,
        dropout=0.1
    ):
        super().__init__()

        # Stack multiple Transformer Encoder Blocks
        self.layers = nn.ModuleList([
            TransformerEncoderBlock(
                embed_dim=embed_dim,
                num_heads=num_heads,
                mlp_dim=mlp_dim,
                dropout=dropout
            )
            for _ in range(depth)
        ])

        # Final LayerNorm
        self.norm = nn.LayerNorm(embed_dim)

    def forward(self, x):

        # Pass tokens through every Transformer block
        for layer in self.layers:
            x = layer(x)

        # Final normalization
        x = self.norm(x)

        return x


# Create our Transformer encoder
transformer_encoder = TransformerEncoder(
    depth=4,
    embed_dim=128,
    num_heads=4,
    mlp_dim=256,
    dropout=0.1
)

# Test it
encoder_output = transformer_encoder(tokens)

print("Input shape:          ", tokens.shape)
print("Encoder output shape: ", encoder_output.shape)
print("Number of blocks:     ", len(transformer_encoder.layers))
```

```python
class ClassificationHead(nn.Module):
    def __init__(self, embed_dim=128, num_classes=3, dropout=0.1):
        super().__init__()

        self.head = nn.Sequential(
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, num_classes)
        )

    def forward(self, x):

        # Take the CLS token
        # x shape: [batch, 197, 128]
        cls_token = x[:, 0]

        # cls_token shape:
        # [batch, 128]

        output = self.head(cls_token)

        # output shape:
        # [batch, 3]

        return output


# Create classification head
classification_head = ClassificationHead(
    embed_dim=128,
    num_classes=3
)

# Test it
class_output = classification_head(encoder_output)

print("Encoder output shape:     ", encoder_output.shape)
print("CLS token shape:          ", encoder_output[:, 0].shape)
print("Classification output:    ", class_output.shape)
```

```python
class CustomVisionTransformer(nn.Module):
    def __init__(
        self,
        image_size=224,
        patch_size=16,
        in_channels=3,
        embed_dim=128,
        num_heads=4,
        mlp_dim=256,
        depth=4,
        num_classes=3,
        dropout=0.1
    ):
        super().__init__()

        # -----------------------------------------
        # 1. Convert image into patch embeddings
        # -----------------------------------------

        self.patch_embedding = PatchEmbedding(
            image_size=image_size,
            patch_size=patch_size,
            in_channels=in_channels,
            embed_dim=embed_dim
        )

        # -----------------------------------------
        # 2. Add CLS token + positional embeddings
        # -----------------------------------------

        self.cls_position = AddCLSAndPosition(
            num_patches=self.patch_embedding.num_patches,
            embed_dim=embed_dim
        )

        # -----------------------------------------
        # 3. Transformer Encoder
        # -----------------------------------------

        self.encoder = TransformerEncoder(
            depth=depth,
            embed_dim=embed_dim,
            num_heads=num_heads,
            mlp_dim=mlp_dim,
            dropout=dropout
        )

        # -----------------------------------------
        # 4. Classification head
        # -----------------------------------------

        self.classification_head = ClassificationHead(
            embed_dim=embed_dim,
            num_classes=num_classes,
            dropout=dropout
        )

    def forward(self, x):

        # Image → patch embeddings
        x = self.patch_embedding(x)

        # Add CLS token + positional information
        x = self.cls_position(x)

        # Transformer encoder
        x = self.encoder(x)

        # Classification
        x = self.classification_head(x)

        return x


# Create YOUR Transformer
custom_vit = CustomVisionTransformer(
    image_size=224,
    patch_size=16,
    in_channels=3,
    embed_dim=128,
    num_heads=4,
    mlp_dim=256,
    depth=4,
    num_classes=3,
    dropout=0.1
)

# Move model to CPU/GPU
custom_vit = custom_vit.to(device)

print(custom_vit)
```

```python
import time
import copy

# Store training history
custom_history = {
    "train_loss": [],
    "train_acc": [],
    "val_loss": [],
    "val_acc": [],
    "lr": []
}

# Keep a copy of the best model
best_custom_model_weights = copy.deepcopy(custom_vit.state_dict())
best_custom_val_loss = float("inf")

# Early stopping settings
patience = 5
epochs_without_improvement = 0

print("Starting Custom Vision Transformer training...")
print("-" * 70)

for epoch in range(num_epochs_custom):

    epoch_start = time.time()

    # ==========================================================
    # TRAINING
    # ==========================================================
    custom_vit.train()

    running_train_loss = 0.0
    correct_train = 0
    total_train = 0

    for images, labels in vit_train_loader:

        images = images.to(device)
        labels = labels.to(device)

        # Clear previous gradients
        custom_optimizer.zero_grad()

        # Forward pass
        outputs = custom_vit(images)

        # Calculate loss
        loss = custom_criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update model weights
        custom_optimizer.step()

        # Statistics
        running_train_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        correct_train += (predictions == labels).sum().item()
        total_train += labels.size(0)

    train_loss = running_train_loss / total_train
    train_acc = correct_train / total_train

    # ==========================================================
    # VALIDATION
    # ==========================================================
    custom_vit.eval()

    running_val_loss = 0.0
    correct_val = 0
    total_val = 0

    with torch.no_grad():

        for images, labels in vit_val_loader:

            images = images.to(device)
            labels = labels.to(device)

            # Forward pass
            outputs = custom_vit(images)

            # Validation loss
            loss = custom_criterion(outputs, labels)

            # Statistics
            running_val_loss += loss.item() * images.size(0)

            predictions = torch.argmax(outputs, dim=1)

            correct_val += (predictions == labels).sum().item()
            total_val += labels.size(0)

    val_loss = running_val_loss / total_val
    val_acc = correct_val / total_val

    # ==========================================================
    # LEARNING RATE
    # ==========================================================
    current_lr = custom_optimizer.param_groups[0]["lr"]

    # Update scheduler
    custom_scheduler.step()

    # ==========================================================
    # SAVE HISTORY
    # ==========================================================
    custom_history["train_loss"].append(train_loss)
    custom_history["train_acc"].append(train_acc)

    custom_history["val_loss"].append(val_loss)
    custom_history["val_acc"].append(val_acc)

    custom_history["lr"].append(current_lr)

    # ==========================================================
    # CHECK FOR BEST MODEL
    # ==========================================================
    if val_loss < best_custom_val_loss:

        best_custom_val_loss = val_loss

        best_custom_model_weights = copy.deepcopy(
            custom_vit.state_dict()
        )

        epochs_without_improvement = 0

        best_marker = " <-- BEST"

    else:

        epochs_without_improvement += 1

        best_marker = ""

    # ==========================================================
    # EPOCH INFORMATION
    # ==========================================================
    epoch_time = time.time() - epoch_start

    print(
        f"Epoch [{epoch+1:02d}/{num_epochs_custom}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"LR: {current_lr:.6f} | "
        f"Time: {epoch_time:.1f}s"
        f"{best_marker}"
    )

    # ==========================================================
    # EARLY STOPPING
    # ==========================================================
    if epochs_without_improvement >= patience:

        print()
        print("Early stopping triggered.")
        print(
            f"Validation loss did not improve for "
            f"{patience} consecutive epochs."
        )

        break

# ==============================================================
# RESTORE BEST MODEL
# ==============================================================

custom_vit.load_state_dict(best_custom_model_weights)

print("-" * 70)
print("Training complete.")
print(f"Best validation loss: {best_custom_val_loss:.4f}")
print("Best model weights restored.")
```
